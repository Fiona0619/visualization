```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Next, make some sample data so that we have something to plot!
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
```

```python

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1)
ax.plot(x,y2)
# fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1,y2)
```

```python
# add legend
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1,label = "Person 1")
ax.plot(x,y2,label = "Person 2")
ax.legend(loc = 'lower right')
```

```python
# add more legend
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1,label = "Person 1")
ax.plot(x,y2,label = "Person 2")
ax.legend(loc = 'lower right',
          frameon = True,
          fontsize = 12,
          ncol = 2,
          shadow = True)
```

```python
# Moving the legend out of the area
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1,label = "Person 1")
ax.plot(x,y2,label = "Person 2")
ax.legend(loc = 'upper left',
          bbox_to_anchor = (1,1)
)
```

```python
# Moving the legend 
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1,label = "Person 1")
ax.plot(x,y2,label = "Person 2")
ax.legend(loc = 'lower left',
          bbox_to_anchor = (1,0.5)
)
```

```python
# Scatter plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1,label = "Person 1")
ax.scatter(x,y2,label = "Person 2")
ax.legend(loc = 'lower right')
ax.text(10,95,"This value is important!")
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1,label = "Person 1")
ax.scatter(x,y2,label = "Person 2")
ax.legend(loc = 'lower right')
ax.text(10,95,"This value is important!",
        ha = 'center',
        color = 'red',
        size = 20)
```

Positioning text annotations
We can do this using transforms
Start by making a simple, empty set of axes

```python
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, ". Data:(1, 5)",           transform=ax.transData    ) #extact location on the (x,y) posotion of theplot
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)",   transform=ax.transAxes    ) # proportion of the axies
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure ) # proportion of the figuer size
```

```python
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, ". Data:(1, 5)",           transform=ax.transData    ) #extact location on the (x,y) posotion of theplot
ax.text(0.6, 0.6, ". Axes:(0.5, 0.1)",   transform=ax.transAxes    ) # proportion of the axies
ax.text(0.9, 0.9, ". Figure:(0.2, 0.2)", transform=fig.transFigure ) # proportion of the figuer size
```

```python
# Annotating with arrows
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
          arrowprops=dict(facecolor='black',shrink=0.05))

##fig.show()
```

```python
# Mordifying arrows
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
          arrowprops=dict(arrowstyle='wedge', color = "hotpink"))
```

Axis Lables

```python
#Removing tick marks and labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

#ax.yaxis.set_major_locator(plt.NullLocator()) 
#ax.xaxis.set_major_formatter(plt.NullFormatter()) 
```

```python
#Removing tick marks and labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.yaxis.set_major_locator(plt.NullLocator()) 
ax.xaxis.set_major_formatter(plt.NullFormatter()) 
```

```python
# Limiting tick marks
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(4)) # limit tick marks to 4
```

```python
# Tick mark intervals
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5)) # interval set as 5
```

```python
#Rotating axis labels
# rotate axis labels and horizontal alignment, use xticks()
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

plt.xticks(rotation=45, ha='right')
```

```python
#Activity: Modifying axis title fonts
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45, ha='right')
plt.xlabel("Shiny New X Axis!",color = 'indigo',font = 'serif')
```

```python
font1 = {'family':'serif','color':'indigo'}
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")


plt.xlabel("Shiny New X Axis!", fontsize = 18, fontdict = font1)
```

```python
font1 = {'family':'serif','color':'indigo'}
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xlabel('Shiny New X Axis!', fontsize = 18, fontdict = font1)
```
