```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy

from PIL import Image
import requests
from io import BytesIO
```

```python
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([170, 100, 90, 120, 50])

# using images from the internet
response = requests.get("https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png")
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
# Now make a basic line plot (reusing our data)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
```

```python
# Then add imshow() to add the image we prepared before
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax_image = fig.add_axes([0.1, # x coordinate (ON FIGURE, NOT AXES)
                        0.11, # y coordinate (ON FIGURE, NOT AXES)
                        0.15, # image width
                        0.25] # image height
                        )
ax_image.imshow(image)
ax_image.axis('off')
```

Practical tips and resources that we can use to improve the accessibility of our data visualizations, especially relating to:
1. Colour
2. Text
3. Image Descriptions
4. Getting our visualizations to the intended audiences

'Good' data visualization :
- include data source
- use an appropriate plot type for comparison
- limits its color scheme

'Bad" data visualization:
- Some of the colours end up looking similar or the same (not accessible to certain people with visual impairments)
- lack of specific labels on the plot(it impossible for individuals with either form of colourblindness to make sense of the data the creator was trying to present)

‘Flattening the Curve':
- this one are an efficient and effective form of communicating public health information, but not to everyone
- Font size

Government of Canada Principles for Accessible Design
The Government of Canada provides four broad goals for what we should try to accomplish with accessible design:
- Make information clear, concise and easy to use
- Present information in predictable, barrier‑free ways
- Provide barrier‑free ways for people to interact with you
- Make sure technology-based products work with assistive technologies and devices


1. COLOR
- Colour is one of the most useful features of many data visualizations
- We have to be aware of the ways that colour can be perceived differently by the audiences of our images
- Colour Blindness Simulator
   -- check contrast: make high contrast
                    tool :online tool, Web Content Accessibility Guidelines
                          Viridis package
-  Moving beyond colour
   -- avoid relying only on colour, where possible
   -- use of patterns or textures in addition to colour on our bars could have made the visualization more accessible to people with visual impairments, even with the current colour scheme

2. TEXT
- Labels and legends
  -- If we want to minimize colours and patterns, data labels (on individual data points or on bars in a bar graph) can make visualizations easier to follow
- Fonts and typefaces: 
  --Certain typefaces and fonts are considered more accessible than others in design and data visualization contexts
- Typeface
   -- generally sans-serif fonts are more accessible for people with cognitive disabilities
   -- avoid overwhelming users by switching between too many different typefaces in one graphic
- Font size and spacing
   -- at least 12 point size or greater
   -- Monospaced text (each character takes the same amount of space) is more widely accessible than proportionally spaced text (each character only takes the space it needs)

3. Describing data visualizations
- Alt-text 
   -- alternate text (alt-text) is used for people using a screen reader or unable to see the image
   -- The most appropriate alt-text can depend on the intended purpose of an image or visualization
   -- Alt-text for graphs and plots of data
   -- Four types of descriptive content ( not equally helpful)
