Multiple viz on one axes object

```python
# Super easy: just call multiple plot methods on the same axes
# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))

# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "indigo")
ax.plot(x, y2,
color = "red")
```

Add error information (error bar etr.)

```python
# First, calculate standard deviation of our data
y2_sd = np.std(y2)
# Then plot our line as before
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
# Then use errorbar() to add in our error line (standard deviation in this case, but could be whatever value you calculated)
ax.errorbar(x, #our x values
        y2, #our y values
        yerr = y2_sd,
        fmt = "none")
```

```python
# Customizing errorbar appearance
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2,
        color = "red")
ax.errorbar(x,
        y2,
        yerr = y2_sd,
        fmt = "none",
        ecolor= "indigo",
        elinewidth= 2,
        capsize = 2,
        capthick= 2 )
```

```python
#If we don’t want to see error bars for every single point, we can specify intervals using errorevery
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2,
        color = "red")
ax.errorbar(x,
        y2,
        yerr = y2_sd,
        fmt = "none",
        ecolor = "indigo",
        elinewidth = 4,
        capsize = 6,
        capthick = 4,
        errorevery= 2 )
```

```python
# the other code 
import numpy as np
import matplotlib.pyplot as plt

x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"]) 
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# Define figure and axes
fig, ax = plt.subplots(figsize=(7, 3))

# Plot the two lines
ax.plot(x, y1, color="indigo", label="Line 1")
ax.plot(x, y2, color="red", label="Line 2")

# Compute standard deviations
y1_sd = np.std(y1)
y2_sd = np.std(y2)

# Convert categorical x-axis to numeric indices
x_numeric = np.arange(len(x))

# Add error bars using numeric indices
ax.errorbar(x_numeric, y2, yerr=y2_sd, fmt="none",
            ecolor="green", elinewidth=4, capsize=6, capthick=4, errorevery=1)
ax.errorbar(x_numeric, y1, yerr=y1_sd, fmt="none",
            ecolor="blue", elinewidth=2, capsize=2, capthick=2)

# Set x-ticks back to categorical labels
ax.set_xticks(x_numeric)
ax.set_xticklabels(x)

# Add legend and show plot
ax.legend()
plt.show()
```
