```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Load data
tips = sns.load_dataset("tips")
print(tips)
```

```python
# basic plot
sns.lineplot(data=tips,       # choose dataset
           x='total_bill',  # define x variable
           y='tip')         # define y variable
```

```python
# Use premade styles to alter the plot
sns.set_style('whitegrid')

tipgraph = sns.lineplot(data=tips,
                        x='total_bill',
                        y='tip')
tipgraph.set(title='Tips vs. Total Bill', 
            xlabel='Total Bill ($)', 
            ylabel='Tip Amount ($)')
```

Plotly
- Plotly syntax is different from matplotlib/seaborn
- Plotly features boxes will show at the up right corner of the plot when moce the mouse at this place

```python
# Import package, make sample data
import plotly.graph_objects as go     # 'go' is 'graph objects’

x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
# make the plot
# First, make our figure, making a container
graph = go.Figure()
# Next, choose our plot type (bar) and our data
graph.add_trace(go.Bar(x=x1, y=y1))
graph.update_layout(
    title="Pirate Scores",
    xaxis_title="Pirates",
    yaxis_title="Score")
# Finally, show our plot!
graph.show()
```

```python
# Customizing plotly graphs
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1, y=y1, mode='markers',    # the variables and style of the plot
    marker=dict(
        size=15,                            # point size
        color='hotpink',                    # point colour
        opacity=1,                          # point transparency, 1 is fully visible while opacity = 0 is completely invisible
        line=dict(width=5, color='purple')  # point outline
    )))

graph.update_layout(
    title='Interactive Pirate Plot',
    xaxis_title='Pirates',
    yaxis_title='Scores',
    width=500, height=500)
```

```python
# Making wordclouds
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
on_bad_lines='skip')
df
```

```python
# join all our text from each row from our quote column into a string

text = " ".join(each for each in df.quote)

# generate our wordcloud image

wordcloud = WordCloud(background_color="white",
colormap = 'inferno').generate(text)

# use matplotlib syntax to put our image in a figure

fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud,                  # remember 'imshow' from when we added pictures to our matplotlib axes
        interpolation='bilinear')   # this line helps smooth our image
ax.axis("off")
plt.show()
# the color scale would be differ 
```

```python
#We can use the matplotlib_venn extension to make venn diagrams
#First import the package
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
#Then define our sets
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
```

```python
#Next, assign our sets to each circle and modify the appearance
diagram = venn2_unweighted([A, B],
                            set_labels = ('Fruits', 'Pies'),
                            set_colors=("blue", "red"),
                            alpha=0.5)
plt.show() 
# By default, the output will only show counts of how many items are in each set
```

```python
# Modyfy

A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])

diagram = venn2_unweighted([A, B],
                            set_labels = ('Fruits', 'Pies'),
                            set_colors=("blue", "red"),
                            alpha=0.5)

diagram.get_label_by_id("10").set_text("n".join(A - B)) # define left circle

diagram.get_label_by_id("11").set_text("n".join(A & B)) #middle

diagram.get_label_by_id("01").set_text("n".join(B - A))  # right
```

```python
# Modyfy 2, wrapp the Texts
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])

diagram = venn2_unweighted([A, B],
                            set_labels = ('Fruits', 'Pies'),
                            set_colors=("blue", "red"),
                            alpha=0.5)

diagram.get_label_by_id("10").set_text("\n".join(A - B)) # define left circle

diagram.get_label_by_id("11").set_text("\n".join(A & B)) # define the middle circle

diagram.get_label_by_id("01").set_text("\n".join(B - A))  # define the right circle
```
